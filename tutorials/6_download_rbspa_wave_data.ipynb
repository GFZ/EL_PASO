# Tutorial: Download Wave Data from RBSP-A

This notebook guides you through downloading and loading wave data from RBSP-A using the **EL PASO framework**.

## Downloading WFR, MAG, Density, and Wave Properties

- **WFR**: Waveform Receiver data
- **MAG**: Magnetometer data  
- **Density**: Plasma density measurements
- **Wave Properties**: Derived wave parameters


````python
import sys
sys.path.append("../")
import el_paso as ep
from datetime import datetime, timezone
from pathlib import Path

# Create save directory
SAVE_PATH = Path('./wave_data/rbspa/')
SAVE_PATH.mkdir(parents=True, exist_ok=True)

# Date range
start_time = datetime(2018, 4, 12, tzinfo=timezone.utc)
end_time = datetime(2018, 4, 12, tzinfo=timezone.utc)

# 1. WFR (Waveform Receiver)
ep.download(
    start_time, end_time,
    save_path=SAVE_PATH,
    download_url='https://cdaweb.gsfc.nasa.gov/pub/data/rbsp/rbspa/l2/emfisis/wfr/spectral-matrix-diagonal/YYYY/',
    file_name_stem="rbsp-a_wfr-spectral-matrix-diagonal_emfisis-l2_YYYYMMDD_.{6}.cdf",
    file_cadence='daily', method='request', skip_existing=True
)

# 2. Density
ep.download(
    start_time, end_time,
    save_path=SAVE_PATH,
    download_url='https://cdaweb.gsfc.nasa.gov/pub/data/rbsp/rbspa/l4/emfisis/density/YYYY/',
    file_name_stem="rbsp-a_density_emfisis-l4_YYYYMMDD_.{7}.cdf",
    file_cadence='daily', method='request', skip_existing=True
)

# 3. Magnetometer
ep.download(
    start_time, end_time,
    save_path=SAVE_PATH,
    download_url='https://cdaweb.gsfc.nasa.gov/pub/data/rbsp/rbspa/l3/emfisis/magnetometer/4sec/sm/YYYY/',
    file_name_stem="rbsp-a_magnetometer_4sec-sm_emfisis-l3_YYYYMMDD_.{6}.cdf",
    file_cadence='daily', method='request', skip_existing=True
)

# 4. Wave Properties
ep.download(
    start_time, end_time,
    save_path=SAVE_PATH,
    download_url='https://cdaweb.gsfc.nasa.gov/pub/data/rbsp/rbspa/l4/emfisis/wna-survey-sheath-corrected-e/YYYY/',
    file_name_stem="rbsp-a_wna-survey-sheath-corrected-e_emfisis-l4_YYYYMMDD_.{6}.cdf",
    file_cadence='daily', method='request', skip_existing=True
)
````
